```python
def linear_search(arr,key):
    for i in range(len(arr)):
        if arr[i]==key:
            return i
    return -1

def binary_search(arr,key):
    low=0
    high=len(arr)-1

    while low<=high:
        mid=(low+high)//2

        if arr[mid]==key:
            return mid
        elif arr[mid]<key:
            low=mid+1
        else:
            high=mid-1

    return -1

arr=list(map(int,input("Enter elements separated by space:").split()))
key=int(input("Enter element to search:"))

position=linear_search(arr,key)

if position!=-1:
    print("Linear_search:element found at position",position)
else:
    print("Linear_search:element not found")
arr.sort()
print("Sorted array:",arr)

position=binary_search(arr,key)

if position!=-1:
    print("Binary_search:element found at position",position+1)
else:
    print("Binary_search:element not found")

    
          
```
